Train Test Split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    heart_df.drop('HeartDisease', axis=1),
    heart_df['HeartDisease'],
    test_size=0.2,
    random_state=42,
    stratify=heart_df['HeartDisease']                #ensures same proportion(like 3:2) of labelled dataset(Y|N) in Train,Test
)
```

Model training

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression        #model
from sklearn.metrics import accuracy_score                 #performance evaluation

solver = ['lbfgs', 'liblinear', 'newton-cg' , 'newton-cholesky', 'sag', 'saga']
best_solver = ''
test_score = np.zeros(6)
for i, n in enumerate(solver):                                    #loop to run all 6
  lr = LogisticRegression(solver=n).fit(x_train, y_train)         #model created
  test_score[i] = lr.score(x_test, y_test)                        #acc
  if lr.score(x_test, y_test) == test_score.max():
    best_solver = n

print(best_solver)
lr = LogisticRegression(solver=best_solver)
lr.fit(x_train, y_train)
lr_pred = lr.predict(x_test)
print(f'LogisticRegression Score: {accuracy_score(y_test, lr_pred)}')
```

```python
import pickle
file = open('LogisticR.pkl', 'wb')    #create a var file, open any pkl, wb - write binary (status)
pickle.dump(lr, file)                 #lr- var in which model is stored, file- file name where we are storing
```

Support vector machine

```python
from sklearn.svm import SVC        #SVClassifier
from sklearn.metrics import f1_score  #performance evaluation matrix

kernels = {'linear':0, 'poly':0, 'rbf':0, 'sigmoid':0}
best = ''    #store name of kernel giving max accuracy
for i in kernels:
  svm = SVC(kernel=i) #instance created
  svm.fit(x_train, y_train) #fit model inside the instance
  yhat = svm.predict(x_test)  #store prediction result
  kernels[i] = f1_score(y_test, yhat, average="weighted") #calculate f1_score of y_test with yhat and initialize average
  if kernels[i] == max(kernels.values()):
    best = i    #max kernel store here

print(best)
svm = SVC(kernel=best)
svm.fit(x_train, y_train)
svm_pred = svm.predict(x_test)
print(f'SVM f1_score kernel({best}): {f1_score(y_test, svm_pred, average="weighted")}')
```

```python
import pickle
file = open('SVM.pkl', 'wb')
pickle.dump(svm, file)
```

Decision tree classifier

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV  #carries all possible combinations for below(ex: md-3,ms-2,ml-2,rs-0)

dtree = DecisionTreeClassifier(class_weight='balanced')
param_grid = {
    'max_depth': [3, 4, 5, 6, 7, 8],
    'min_samples_split': [2, 3, 4],
    'min_samples_leaf': [1, 2, 3, 4],
    'random_state': [0, 42]
}
grid_search = GridSearchCV(dtree, param_grid, cv=5)   #pass model name dtree inside the func and cv(cross validation parameter) as 5
grid_search.fit(x_train, y_train)     #fit training dataset
Ctree = DecisionTreeClassifier(**grid_search.best_params_, class_weight='balanced')    #best parameter list  from **grid_search, access by best_params , store the best decision tree model in Ctree
Ctree.fit(x_train, y_train)
dtc_pred = Ctree.predict(x_test)
print("DecisionTrees's Accuracy: ", accuracy_score(y_test, dtc_pred))
```

```python
import pickle
file = open('DecisionTree.pkl', 'wb')
pickle.dump(Ctree, file)
```

Random forest classifier

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score

rfc = RandomForestClassifier()
param_grid = {
    'n_estimators': [50, 100, 150, 500],
    'max_features': ['sqrt', 'log2', None],
    'max_depth': [3, 6, 9, 19],
    'max_leaf_nodes': [3,6,9],
}
grid_search = GridSearchCV(rfc, param_grid)
grid_search.fit(x_train, y_train)
rfctree = RandomForestClassifier(**grid_search.best_params_)
rfctree.fit(x_train, y_train)
rfc_pred = rfctree.predict(x_test)
print("RandomForestClassifier's Accuracy: ", accuracy_score(y_test, rfc_pred))
```

```python
import pickle
file = open('RandomForest.pkl', 'wb')
pickle.dump(rfctree, file)
```
